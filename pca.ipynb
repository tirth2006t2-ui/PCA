```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import time
from sklearn.datasets import load_iris
```

```python
lr=load_iris()
```

```python
lr.target
```

```python
df = pd.DataFrame(lr.data, columns=lr.feature_names)
df["target"] = lr.target
```

```python
df.sample()
```

```python
k=KNeighborsClassifier()
```

```python
xtr,xts,ytr,yts=train_test_split(df.drop(columns=["target"]),df.target)
```

```python
k.fit(xtr,ytr)
k.score(xts,yts)
```

```python
s=StandardScaler()
xtr=s.fit_transform(xtr)
xts=s.transform(xts)
```

```python
p=PCA(n_components=2)
xtr=p.fit_transform(xtr)
xts=p.transform(xts)
```

```python
k=KNeighborsClassifier()
```

```python
k.fit(xtr,ytr)
k.score(xts,yts)
```
